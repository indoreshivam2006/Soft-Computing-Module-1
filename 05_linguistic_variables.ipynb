## Practical 5: Fuzzy Linguistic Variables (Low, Medium, High)

### Aim
To create the linguistic variable *Temperature* with the terms Low, Medium and High, and to fuzzify a crisp input.

### Theory
A linguistic variable is a variable whose values are words, such as Low, Medium or High, each represented by a fuzzy set. Neighbouring terms overlap, so one crisp value can belong to two terms at the same time. Fuzzification is the step that converts a crisp input into its membership in each term.

### Implementation


```python
import numpy as np
import matplotlib.pyplot as plt
import skfuzzy as fuzz

# 1. Universe of discourse for temperature (0 to 40 C)
temp = np.arange(0, 40.01, 0.1)

# 2. Membership functions for the three terms
low = fuzz.trapmf(temp, [0, 0, 10, 20])
medium = fuzz.trimf(temp, [10, 20, 30])
high = fuzz.trapmf(temp, [20, 30, 40, 40])
```

```python
# 3. Fuzzify a crisp input
crisp_value = 24.0
print(f"Crisp temperature = {crisp_value} C")
print(f"  Low    : {fuzz.interp_membership(temp, low, crisp_value):.2f}")
print(f"  Medium : {fuzz.interp_membership(temp, medium, crisp_value):.2f}")
print(f"  High   : {fuzz.interp_membership(temp, high, crisp_value):.2f}")
```

### Visualization


```python
plt.figure(figsize=(8, 4))
plt.plot(temp, low, label="Low", linewidth=2)
plt.plot(temp, medium, label="Medium", linewidth=2)
plt.plot(temp, high, label="High", linewidth=2)
plt.axvline(crisp_value, color="black", linestyle="--", label=f"input = {crisp_value}")
plt.xlabel("Temperature (C)")
plt.ylabel("Membership")
plt.title("Linguistic variable: Temperature")
plt.ylim(-0.05, 1.1)
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

### Observation
- An input of 24 C has membership 0.00 in Low, 0.60 in Medium and 0.40 in High.
- The term Medium is strongest here, but the input is still partly High, because the two terms overlap between 20 C and 30 C.

### Conclusion
A linguistic variable with three overlapping terms was defined, and a crisp temperature was fuzzified into its membership in each term.
