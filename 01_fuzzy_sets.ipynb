## Practical 1: Fuzzy Sets using Membership Values

### Aim
To create and visualize a simple fuzzy set using membership values in Python.

### Theory
In a classical (crisp) set, an element either belongs to the set or it does not. A fuzzy set $A$ on a universe $X$ is defined by a membership function $\mu_A : X \to [0, 1]$, where $\mu_A(x)$ is the degree to which $x$ belongs to $A$.

Here the fuzzy set *Young* is defined over a few ages with explicit membership values.

### Implementation


```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Universe of discourse (ages) and membership values for "Young"
ages = [10, 20, 30, 40, 50, 60, 70]
mu_young = [1.0, 1.0, 0.7, 0.4, 0.1, 0.0, 0.0]

# 2. Store the fuzzy set as a dictionary: element -> membership
young = dict(zip(ages, mu_young))

# 3. Membership values must lie between 0 and 1
assert all(0 <= m <= 1 for m in mu_young)

print("Fuzzy set 'Young':")
for age, m in young.items():
    print(f"  Age {age}: {m:.2f}")
```

```python
# 4. Support (membership > 0) and alpha-cut at alpha = 0.5
support = [a for a, m in young.items() if m > 0]
alpha_cut = [a for a, m in young.items() if m >= 0.5]

print("Support        :", support)
print("Alpha-cut (0.5):", alpha_cut)
```

### Visualization


```python
plt.figure(figsize=(7, 4))
plt.bar(ages, mu_young, width=6, color="steelblue", edgecolor="black")
plt.plot(ages, mu_young, "ro--")
plt.ylim(0, 1.1)
plt.xlabel("Age")
plt.ylabel("Membership value")
plt.title("Fuzzy Set: Young")
plt.grid(axis="y", alpha=0.3)
plt.show()
```

### Observation
- Age 30 has membership 0.70, so it is young to a degree, not fully in or out of the set.
- The support is ages 10 to 50. The alpha-cut at 0.5 keeps ages 10, 20 and 30.

### Conclusion
A fuzzy set was created from membership values, its support and alpha-cut were computed, and the membership curve was plotted.
