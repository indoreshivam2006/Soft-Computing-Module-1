## Practical 2: Triangular, Trapezoidal and Gaussian Membership Functions

### Aim
To create and plot triangular, trapezoidal and Gaussian membership functions using scikit-fuzzy.

### Theory
- **Triangular** $(a, b, c)$: rises linearly from $a$ to a peak at $b$, then falls to $c$.
- **Trapezoidal** $(a, b, c, d)$: rises from $a$ to $b$, stays at 1 until $c$, then falls to $d$.
- **Gaussian** (mean $m$, spread $\sigma$): $\mu(x) = e^{-(x-m)^2 / (2\sigma^2)}$, a smooth bell curve.

### Implementation


```python
import numpy as np
import matplotlib.pyplot as plt
import skfuzzy as fuzz

# 1. Universe of discourse
x = np.arange(0, 10.01, 0.01)

# 2. Create the three membership functions
tri = fuzz.trimf(x, [2, 5, 8])               # a=2, b=5, c=8
trap = fuzz.trapmf(x, [1, 3, 7, 9])          # a=1, b=3, c=7, d=9
gauss = fuzz.gaussmf(x, mean=5, sigma=1.2)   # centre 5, spread 1.2
```

```python
# 3. Membership values at a few points
for point in [3, 5, 7]:
    print(f"x = {point}: triangular = {fuzz.interp_membership(x, tri, point):.3f}, "
          f"trapezoidal = {fuzz.interp_membership(x, trap, point):.3f}, "
          f"gaussian = {fuzz.interp_membership(x, gauss, point):.3f}")
```

### Visualization


```python
fig, axes = plt.subplots(3, 1, figsize=(7, 8), sharex=True)
names = ["Triangular", "Trapezoidal", "Gaussian"]
curves = [tri, trap, gauss]

for ax, curve, name in zip(axes, curves, names):
    ax.plot(x, curve, linewidth=2)
    ax.set_title(name)
    ax.set_ylabel("mu(x)")
    ax.set_ylim(-0.05, 1.1)
    ax.grid(alpha=0.3)

axes[-1].set_xlabel("x")
plt.tight_layout()
plt.show()
```

### Observation
- At $x = 3$ the triangular function gives 0.333, while the trapezoidal function gives 1.000 because 3 is on its flat top.
- The Gaussian gives 0.249 at $x = 3$ and never drops exactly to 0.

### Conclusion
Triangular, trapezoidal and Gaussian membership functions were implemented and plotted with scikit-fuzzy.
